## Imbalance Dataset:
a dataset with an unequal class distribution

```python
# importing the dependencies
import numpy as np
import pandas as pd
```

```python
# loading a dataset to a pandas DataFrame
creditCard_data = pd.read_csv('credit_data.csv')
```

```python
creditCard_data.head()
```

```python
creditCard_data['Class'].value_counts()
```

This is Highly Imbalanced Dataset

0 -> Legit Transactions

1 -> Fraudulent Transactions

```python
legit = creditCard_data[creditCard_data.Class == 0]
fraud = creditCard_data[creditCard_data.Class == 1]
```

```python
print(legit.shape)
print(fraud.shape)
```

Under-sampling

Build a simple dataset containing similar distribution of legit & fraudulent Transactions

number of fraudulent Transaction -> 245

```python
legit_sample = legit.sample(250)
```

```python
legit_sample
```

Concatenate the Two Dataframes

```python
new_data = pd.concat([legit_sample,fraud], axis=0)
```

```python
new_data
```

```python
new_data['Class'].value_counts()
```
