Importing the Dependencies

```python
import numpy as np
import pandas as pd
# from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.datasets import fetch_openml
```

loading the dataset

```python
sonar = fetch_openml(name='sonar',version=1,as_frame=True)
```

```python
X = sonar.data
Y = sonar.target
```

```python
X
```

```python
# number of rows and columns
X.shape
```

```python
# get statistical measures of the data
X.describe()
```

```python
# checking the values of each
Y.value_counts()
```

```python
# get the mean of each of the 60 features.
X.groupby(Y).mean()
```

Spliting the data into training and tes data

```python
x_train, x_test , y_train, y_test = train_test_split(X,Y, test_size=0.1, stratify=Y,random_state=1)
```

```python
print(X.shape,x_train.shape,x_test.shape)
```

```python
print(Y.shape,y_train.shape,y_test.shape)
```

Model Training -> Logistic Regression

```python
model = LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
## Model Evaluation
train_prediction = model.predict(x_train)
train_prediction_accuracy = accuracy_score(y_train, train_prediction)


```

```python
print('Accuracy training data: ',train_prediction_accuracy)
```

```python
test_prediction = model.predict(x_test)
test_prediction_accuracy = accuracy_score(test_prediction,y_test)
```

```python
print('Accuracy testing data: ',test_prediction_accuracy)
```
