```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
from sklearn.datasets import fetch_openml
```

```python
diabetes = fetch_openml(name='diabetes',as_frame=True)
```

```python
X = diabetes.data
Y = diabetes.target
```

```python
X.shape
```

getting the statictical measures of the data

```python

X.describe()
```

```python
Y.value_counts()
```

```python
X.groupby(Y).mean()
```

Split into Training and Test data

```python
x_train,x_test,y_train,y_test = train_test_split(X,Y,test_size=0.2,stratify=Y,random_state=2)
```

Data Standardization

```python
scaler = StandardScaler()
```

```python
scaler.fit(x_train)
```

```python
x_train = scaler.transform(x_train)
x_test = scaler.transform(x_test)
```

Training the model

```python
classifier_model = svm.SVC(kernel='linear')
```

```python
# training the super vector machine classifier
classifier_model.fit(x_train,y_train)
```

```python
train_prediction = classifier_model.predict(x_train)
test_prediction = classifier_model.predict(x_test)
```

Model Evaluation

```python
# the accuracy of the training model or evaluating the model
train_prediction_accuracy = accuracy_score(train_prediction,y_train)
```

```python
print('The Accuracy Score of training Data: ',train_prediction_accuracy)
```

```python
# testing the accuracy of the test model or evaluating the model
test_prediction_accuracy = accuracy_score(test_prediction,y_test)
```

```python
print('The Accuracy Score of testing Data: ',test_prediction_accuracy)
```

## Making a Predictive System

```python
input = (1.0,95.0,74.0,21.0,73.0,25.9,0.673,36.0)


# changing the input data to a numpy array
input_as_array = np.asarray(input)

# reshape the np array
input_reshape = input_as_array.reshape(1,-1)

#standardize the input data
std_data = scaler.transform(input_reshape)

predict = classifier_model.predict(std_data)

if predict[0] == 'tested_positive':
  print('The person is diabetic')
else:
  print('The person is not diabetic')
```
