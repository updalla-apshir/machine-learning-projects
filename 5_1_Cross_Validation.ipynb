importing the dependencies

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score
from sklearn.datasets import fetch_openml
```

import the models

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
heart = fetch_openml(name='heart-disease', as_frame=True,version=1)
```

```python
heart_data = heart.data
```

```python
heart_data
```

```python
# split the labels and features
X = heart_data.drop('target',axis=1)
Y = heart_data['target']
```


#**Train Test Split**

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y , stratify = Y, test_size=0.2)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

Comparing the performace of the models

```python
# load the models

models = [LogisticRegression(max_iter=1000),SVC(kernel='linear'),KNeighborsClassifier(),RandomForestClassifier()]
```

```python
def compare_models_train_test():

  for model in models:

    # train the model
    model.fit(X_train,Y_train)

    # evaluating the model
    test_data_prediction = model.predict(X_test)

    accuracy = accuracy_score(test_data_prediction,Y_test)

    print('Accuracy score of the ',model,' = ',accuracy)
```

```python
compare_models_train_test()
```

```python
# Accuracy score of the  LogisticRegression(max_iter=1000)  =  0.8032786885245902
# Accuracy score of the  SVC(kernel='linear')  =  0.7868852459016393
# Accuracy score of the  KNeighborsClassifier()  =  0.6721311475409836
# Accuracy score of the  RandomForestClassifier()  =  0.8524590163934426------> highest
```

#**Cross Validation**

Logistic Regression

```python
cross_score_l = cross_val_score(LogisticRegression(max_iter=1000),X,Y,cv=5)
print(cross_score_l)
print('average logistic regression model is: ' , cross_score_l.mean())
```

Supper Vector Machine

```python
cross_score_s = cross_val_score(SVC(kernel='linear'),X,Y,cv=5)
print(cross_score_s)
print('average Supper Vector Machine model is: ' , cross_score_s.mean())
```

K Nearest Neighbors Classifier

```python
cross_score_k = cross_val_score(KNeighborsClassifier(),X,Y,cv=5)
print(cross_score_k)
print('average K Nearest Neighbors Classifier Machine model is: ' , cross_score_k.mean())
```

Random Forest Classifier

```python
cross_score_r = cross_val_score(RandomForestClassifier(),X,Y,cv=5)
print(cross_score_r)
print('average Random Forest Classifier Machine model is: ' , cross_score_r.mean())
```

```python
def compare_models_cross_validation():

  for model in models:

    cross_score = cross_val_score(model,X,Y,cv=5)
    print(cross_score)
    print('average ',model,' model is: ' ,round( cross_score.mean()*100,2), '%')
```

```python
compare_models_cross_validation()
```
